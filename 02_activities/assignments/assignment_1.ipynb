### Classification using KNN

Let's set up our workspace and use the **Wine dataset** from `scikit-learn`. This dataset contains 178 wine samples with 13 chemical features, used to classify wines into different classes based on their origin.

The **response variable** is `class`, which indicates the type of wine. We'll use all of the chemical features to predict this response variable.

```python
# Import standard libraries
import pandas as pd
import numpy as np
import random
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import recall_score, precision_score
from sklearn.model_selection import cross_validate
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score
```

```python
from sklearn.datasets import load_wine

# Load the Wine dataset
wine_data = load_wine()

# Convert to DataFrame
wine_df = pd.DataFrame(wine_data.data, columns=wine_data.feature_names)

# Bind the 'class' (wine target) to the DataFrame
wine_df['class'] = wine_data.target

# Display the DataFrame
wine_df
```

#### **Question 2:** 
#### Standardization and data-splitting

Next, we must preform 'pre-processing' or 'data munging', to prepare our data for classification/prediction. For KNN, there are three essential steps. A first essential step is to 'standardize' the predictor variables. We can achieve this using the scaler method, provided as follows:

```python
# Select predictors (excluding the last column)
predictors_standardized = wine_df.copy()
columns_to_exclude = ['class']
columns_to_scale = predictors_standardized.columns.difference(columns_to_exclude)

# Standardize the predictors
scaler = StandardScaler()
predictors_standardized[columns_to_scale] = scaler.fit_transform(wine_df[columns_to_scale])

# Display the head of the standardized predictors
print(predictors_standardized.head())
```

(i) Why is it important to standardize the predictor variables?

> Your answer here...
Standardization prevents variables with large values (like in 'proline') from having a much greater influence than variables with small values (like in 'nonflavanoid_phenols'). By standardizing the variables to be relative values each variable has an equal weighting. 

(ii) Why did we elect not to standard our response variable `Class`?

> Your answer here...
Although the 'class' variable is an integer the values in it reflect categories of wines. We do not want to standardize our response variables like 'class' because it will distort the values of the real categories. 

(iii) A second essential step is to set a random seed. Do so below (Hint: use the random.seed function). Why is setting a seed important? Is the particular seed value important? Why or why not?

> Your answer here...
Setting a seed helps with consistency and reproducability. This allows us to randomly assign data to either the train or test set while still being able to reproduce the results at a later time. If we didn't have this ability we would have to either subset the data in a non-random and potentially biased way or have a scenario where each time you run the code you get different results.

The particular seed value is arbitrary as it just sets the start point of the random sequence so which sequence of random values is used does not matter. However, to reproduce previous results the seed needs to be consistent. 

(iv) A third essential step is to split our standardized data into separate training and testing sets. We will split into 75% training and 25% testing. The provided code randomly partitions our data, and creates linked training sets for the predictors and response variables. 

Extend the code to create a non-overlapping test set for the predictors and response variables.

```python
# set a seed for reproducibility
np.random.seed(123)

# split the data into a training and testing set. hint: use train_test_split !

# Your code here ...

wine_train, wine_test = train_test_split(
    predictors_standardized, train_size=0.75, stratify=wine_df["class"]
)
```

#### **Question 3:**
#### Model initialization and cross-validation
We are finally set to fit the KNN model. 


Perform a grid search to tune the `n_neighbors` hyperparameter using 10-fold cross-validation. Follow these steps:

1. Initialize the KNN classifier using `KNeighborsClassifier()`.
2. Define a parameter grid for `n_neighbors` ranging from 1 to 50.
3. Implement a grid search using `GridSearchCV` with 10-fold cross-validation to find the optimal number of neighbors.
4. After fitting the model on the training data, identify and return the best value for `n_neighbors` based on the grid search results.

```python
# Initilize knn
knn = KNeighborsClassifier()
knn
```

```python
# Set parameter grid from 1 to 50
parameter_grid = {
    "n_neighbors": range(1, 50, 1),
}
```

```python
# Run grid search with 10-fold cross validation
wine_tune_grid = GridSearchCV(
    estimator=knn,
    param_grid=parameter_grid,
    cv=10
)
```

```python
wine_train
```

```python
# Fit the model on the grid predicting the response variable on the validation sets using the information
# in the predictor variables 
wine_tune_grid.fit(
    wine_train.iloc[:, :-1],
    wine_train.iloc[:, -1:]
)
```

```python
# Add results to a dataframe
accuracy_grid = pd.DataFrame(wine_tune_grid.cv_results_)
accuracy_grid
```

```python
# Plot the accuracy by the number of neighbours 

plt.figure(figsize=(10, 6))

plt.plot(accuracy_grid['param_n_neighbors'], accuracy_grid['mean_test_score'], '-o', color='blue')

plt.xlabel('Number of Neighbors')
plt.ylabel('Accuracy estimate')
plt.title('K-Nearest Neighbors Performance')
plt.tight_layout()
plt.show()

# The accuracy peaks at k=7 with an accuracy of 0.977473	
```

```python
wine_tune_grid.best_params_
```

#### **Question 4:**
#### Model evaluation

Using the best value for `n_neighbors`, fit a KNN model on the training data and evaluate its performance on the test set using `accuracy_score`.

```python
# Your code here...
# Initilize the new knn model with the best number of neighbors (7)
knn_optimized = KNeighborsClassifier(n_neighbors=7)
knn_optimized
```

```python
# Fit the model to the training data
knn_optimized.fit(
    wine_train.iloc[:, :-1],
    wine_train.iloc[:, -1:]
)
```

```python
wine_test
```

```python
# Create a new ;predicted column in the test data which gives the predited class
wine_test["predicted"] = knn_optimized.predict(wine_test.iloc[:, :-1])
wine_test[["class", "predicted"]]
```

```python
# Accuracy of the new model
knn_optimized.score(
    wine_test.iloc[:, :-2],
    wine_test.iloc[:, -2:-1]
)

# The model correctly predicts the class of wine 93.3% of the time.
# It seems from this that wine class can be predicted fairly accurately
# from the available predictors. Therefore these classes must represent
# real differences in the wines
```

```python
# Can run a confusion matrix
pd.crosstab(
    wine_test['class'],
    wine_test['predicted']
)

# It seems like all three classes are distinct
```
